````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
try:
    make_safe_order('X', 'buy', -5, 1); raise SystemExit('deberia fallar')
except ValueError:
    pass
assert make_safe_order('X', 'buy', 100, 1)['price'] == 100
print('ok')
````
