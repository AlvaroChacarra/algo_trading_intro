lo impreso; el nombre de módulo modela su primera importación.
Observa `broken_source` como `'demo'` y `guarded_source` como `'demo'` y
`'__main__'`, guardando `broken_output`, `import_output` y `direct_output`.
Deben resultar `'ARRANCANDO BACKTEST'`, `''` y `'ARRANCANDO BACKTEST'`: identifica la guardia que cambia el resultado.


```python
import io
from contextlib import redirect_stdout

# Helper proporcionado: recibe código y nombre; devuelve su salida como texto.
def observe_code(source, run_name):
    output = io.StringIO()
    with redirect_stdout(output):
        exec(source, {'__name__': run_name})
    return output.getvalue().strip()

broken_source = """def main():
    print('ARRANCANDO BACKTEST')

main()
"""

guarded_source = """def main():
    print('ARRANCANDO BACKTEST')

if __name__ == '__main__':
    main()
"""
broken_output = None
import_output = None
direct_output = None
```
