````python
class EmptyBookError(Exception):
    pass

def top(book):
    pass
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
try:
    top([]); raise SystemExit('deberia fallar')
except EmptyBookError:
    pass
assert top([{'id':1}]) == {'id':1}
print('ok')
````

<details>
<summary>💡 Ver solución</summary>

```python
class EmptyBookError(Exception):
    pass

def top(book):
    if not book:
        raise EmptyBookError('el libro está vacío')
    return book[0]
```
