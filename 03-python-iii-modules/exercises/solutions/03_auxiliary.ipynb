### Valida y captura el error esperado

Completa `check_size(size)`: devuelve tamaños positivos y lanza `ValueError`
con el valor rechazado si `size <= 0`.
En `make_safe_order`, rechaza también precios no positivos con un mensaje que
incluya el precio y usa `check_size` antes de devolver el diccionario.
Completa `safe_order` para devolver esa orden o `None` capturando únicamente `ValueError`.


````python
def check_size(size):
    if size <= 0:
        raise ValueError(f'size inválido: {size}')
    return size

def make_safe_order(price, size):
    if price <= 0:
        raise ValueError(f'price inválido: {price}')
    return {'price': price, 'size': check_size(size)}

def safe_order(price, size):
    try:
        return make_safe_order(price, size)
    except ValueError:
        return None
````

````python
print(safe_order(100, 0.5))
print(safe_order(100, 0))
print(safe_order(0, 1))
try:
    check_size(-0.5)
except ValueError as error:
    print(error)
````

**Resultado esperado:** `{'price': 100, 'size': 0.5}`, `None`, `None`; por último un mensaje que incluya `-0.5`.
