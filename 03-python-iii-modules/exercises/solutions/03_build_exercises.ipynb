### Reutilizar tu módulo

Pega tu función `best_prices` de L2 en el hueco indicado y completa `describe(orders)` usando esa función.

Devuelve un diccionario con `mid` y `spread`, aplicando las fórmulas de L1 al par recibido. Si falta algún lado, ambos valores serán `None`.

Ejecuta los cuatro casos. Al terminar, llevarás estas mismas funciones a `mi_main.py`: el notebook permite investigar por celdas; el archivo permite ejecutar el programa e importar sus funciones.


```python
orders = [
    {'side': 'buy', 'price': 98, 'size': 1},
    {'side': 'sell', 'price': 103, 'size': 2},
    {'side': 'buy', 'price': 99, 'size': 4},
    {'side': 'sell', 'price': 101, 'size': 1},
]
def best_prices(orders):
    buys = [order['price'] for order in orders if order['side'] == 'buy']
    sells = [order['price'] for order in orders if order['side'] == 'sell']
    bid = max(buys) if buys else None
    ask = min(sells) if sells else None
    return bid, ask

def describe(orders):
    bid, ask = best_prices(orders)
    if bid is None or ask is None:
        return {'mid': None, 'spread': None}
    return {'mid': (bid + ask) / 2, 'spread': ask - bid}
```

```python
print('ambos lados:', describe(orders))
print('solo compras:', describe([{'side': 'buy', 'price': 99}]))
print('solo ventas:', describe([{'side': 'sell', 'price': 101}]))
print('vacío:', describe([]))
```

**Resultado esperado:** Ambos lados: `{'mid': 100, 'spread': 2}`. En los otros tres casos: `{'mid': None, 'spread': None}`.
