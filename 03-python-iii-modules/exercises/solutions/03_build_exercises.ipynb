### 1 · Del par a las métricas

Completa `metrics(bid, ask)`: si falta un lado devuelve ambas métricas como None; en otro caso calcula mid y spread. Predice el resultado para 99 y 101 antes de ejecutar.

**Base proporcionada.** Funciones canónicas de L2 y las mismas cuatro órdenes: best_prices devuelve (bid, ask), con None cuando falta un lado.

```python
orders = [
    {'side': 'buy', 'price': 98, 'size': 1},
    {'side': 'sell', 'price': 103, 'size': 2},
    {'side': 'buy', 'price': 99, 'size': 4},
    {'side': 'sell', 'price': 101, 'size': 1},
]
def buy_prices(orders):
    prices = []
    for order in orders:
        if order['side'] == 'buy':
            prices.append(order['price'])
    return prices


def best_bid(orders):
    prices = buy_prices(orders)
    if not prices:
        return None
    return max(prices)


def sell_prices(orders):
    prices = []
    for order in orders:
        if order['side'] == 'sell':
            prices.append(order['price'])
    return prices

def best_ask(orders):
    prices = sell_prices(orders)
    if not prices:
        return None
    return min(prices)


def best_prices(orders):
    return best_bid(orders), best_ask(orders)
```

```python
def metrics(bid, ask):
    if bid is None or ask is None:
        return {'mid': None, 'spread': None}
    return {'mid': (bid + ask) / 2, 'spread': ask - bid}
```

```python
print('métricas:', metrics(99,101))
```

**Resultado:** mid=100 y spread=2.

### 2 · Describe con tus funciones

Completa `describe`: llama una vez a best_prices y pasa su par a tu función metrics. Ejecuta la consulta sobre las mismas cuatro órdenes.

```python
def describe(orders):
    bid, ask = best_prices(orders)
    return metrics(bid, ask)
```

```python
print('ambos lados:', describe(orders))
```

**Resultado:** mid=100 y spread=2; las órdenes no cambian.
