### Solución guiada

```python
from exchange import Market, Backtest, Side
from exchange.strategies import VWAPStrategy
def run(p):
    r = Backtest(Market.sample(), VWAPStrategy('BTCUSDT', Side.BUY, 1.0, 5, profile=p)).run()
    return sum(f.size for f in r.fills)
a = run([2,2,2,2,2]); b = run([1,1,1,1,1])
```
