### 1 · Consulta el mid

Completa el hueco 1 de `OrderBook.mid`: usa los mejores precios y devuelve `None` si falta un lado. El constructor ordena Level; `imbalance` ya fue explicado en la presentación.

**Base proporcionada.** Level(price, size), presentado en §1, representa una fila del libro; no es una tupla.

```python
class Level:
    def __init__(self, price, size):
        self.price = price
        self.size = size
```

```python
class OrderBook:
    def __init__(self, bids, asks):
        self.bids = sorted(bids, key=lambda level: -level.price)
        self.asks = sorted(asks, key=lambda level: level.price)

    @property
    def mid(self):
        if not self.bids or not self.asks:
            return None
        return (self.bids[0].price + self.asks[0].price) / 2

    def imbalance(self, levels=1):
        bid = sum(level.size for level in self.bids[:levels])
        ask = sum(level.size for level in self.asks[:levels])
        return (bid - ask) / (bid + ask) if bid + ask else None
```

```python
book = OrderBook([Level(98, 1), Level(99, 2)], [Level(101, 3)])
print('mid:', book.mid)
print('sin asks:', OrderBook([Level(99, 2)], []).mid)
```

**Resultado:** `100.0` y `None`.

### Tu cartera · una sola celda editable

Completa los huecos 2–4 siguiendo los pasos de debajo. Cada cambio de clase requiere ejecutarla y crear instancias nuevas: las antiguas conservan la definición antigua.

**Base proporcionada.** Fill de L4: side, price, size; cash_flow devuelve importe negativo al comprar y positivo al vender.

```python
from math import isfinite

fill_record = {'side': 'buy', 'price': 100, 'size': 2}
class Fill:
    def __init__(self, side, price, size):
        valid_numbers = all(isfinite(x) and x > 0 for x in (price, size))
        if side not in ('buy', 'sell') or not valid_numbers:
            raise ValueError('fill requiere lado válido y valores positivos finitos')
        self.side = side
        self.price = price
        self.size = size

    def cash_flow(self):
        sign = -1 if self.side == 'buy' else 1
        return sign * self.price * self.size
```

```python
class PositionTracker:
    def __init__(self, cash=0.0):
        self.cash = cash
        self.position = 0.0

    def apply_fill(self, fill):
        self.cash += fill.cash_flow()
        self.position += fill.size if fill.side == 'buy' else -fill.size

    def equity(self, mark):
        return self.cash + self.position * mark
```

#### 2 · Lleva el fill a caja

Completa el hueco 2 con `fill.cash_flow()`. Ejecuta la celda de la clase y esta observación; predice la caja después de comprar.

```python
tracker = PositionTracker(cash=1000)
buy = Fill('buy', 101, 2)
tracker.apply_fill(buy)
print('caja:', tracker.cash)
```

**Resultado:** `798`.
