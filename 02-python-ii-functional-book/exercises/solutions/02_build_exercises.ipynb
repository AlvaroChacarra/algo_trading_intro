### 1 · Extrae las compras

Completa el recorrido de `buy_prices`: si la orden es buy, añade su precio. Predice la lista antes de ejecutar.

**Base proporcionada.** Las cuatro órdenes de la presentación y de L1: side, price y size.

```python
orders = [
    {'side': 'buy', 'price': 98, 'size': 1},
    {'side': 'sell', 'price': 103, 'size': 2},
    {'side': 'buy', 'price': 99, 'size': 4},
    {'side': 'sell', 'price': 101, 'size': 1},
]
```

```python
def buy_prices(orders):
    prices = []
    for order in orders:
        if order['side'] == 'buy':
            prices.append(order['price'])
    return prices
```

```python
print('compras:', buy_prices(orders))
```

**Resultado:** `[98, 99]`; conserva el orden de llegada.

<a id="L02-P03"></a>

### 2 · Elige el bid

Completa `best_bid` usando tu `buy_prices`. `max` elige el mayor; si la lista está vacía, conserva la ausencia con `None`.

```python
def best_bid(orders):
    prices = buy_prices(orders)
    if not prices:
        return None
    return max(prices)
```

```python
print('bid:', best_bid(orders))
```

**Resultado:** `99`.

<a id="L02-P04"></a>

### 3 · Repite para ventas

Construye `sell_prices` con el mismo recorrido y `best_ask` con el mínimo. Mantén `None` cuando no haya ventas.

```python
def sell_prices(orders):
    prices = []
    for order in orders:
        if order['side'] == 'sell':
            prices.append(order['price'])
    return prices

def best_ask(orders):
    prices = sell_prices(orders)
    if not prices:
        return None
    return min(prices)
```

```python
print('ventas:', sell_prices(orders))
print('ask:', best_ask(orders))
```

**Resultado:** `[103, 101]` y `101`.

### 4 · Reúne tus dos consultas

Completa `best_prices` llamando a tus dos funciones. Devuelve el par en orden `(bid, ask)` y comprueba que las cuatro órdenes siguen intactas.

```python
def best_prices(orders):
    return best_bid(orders), best_ask(orders)
```

```python
print('bid, ask:', best_prices(orders))
print('órdenes:', orders)
```

**Resultado:** `(99, 101)` y las cuatro órdenes originales.
