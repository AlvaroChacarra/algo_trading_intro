### Variante 1 · Una compra mejor

Crea otra lista añadiendo una compra a 100. Predice y consulta el nuevo bid sin cambiar `orders`.

**Base proporcionada.** Cierre canónico del principal: filtros → best_bid/best_ask → best_prices; mismas cuatro órdenes.

```python
orders = [
    {'side': 'buy', 'price': 98, 'size': 1},
    {'side': 'sell', 'price': 103, 'size': 2},
    {'side': 'buy', 'price': 99, 'size': 4},
    {'side': 'sell', 'price': 101, 'size': 1},
]
def buy_prices(orders):
    prices = []
    for order in orders:
        if order['side'] == 'buy':
            prices.append(order['price'])
    return prices


def best_bid(orders):
    prices = buy_prices(orders)
    if not prices:
        return None
    return max(prices)


def sell_prices(orders):
    prices = []
    for order in orders:
        if order['side'] == 'sell':
            prices.append(order['price'])
    return prices

def best_ask(orders):
    prices = sell_prices(orders)
    if not prices:
        return None
    return min(prices)


def best_prices(orders):
    return best_bid(orders), best_ask(orders)
```

```python
better = orders + [{'side': 'buy', 'price': 100, 'size': 1}]
```

```python
print(best_prices(better))
print(best_prices(orders))
```

**Resultado:** `(100, 101)` y `(99, 101)`.

### Ampliación · Filtra sin cambiar la lista

Implementa `cancel_order(book, order_id)` con `for` y `append`, y
`cancel_compact` con una comprensión: ambas devuelven una lista nueva sin ese id,
conservando el orden y los diccionarios originales.
Guarda en `new_list` la cancelación del id `1` y en `same_order` si su único
elemento es el mismo objeto que `book[1]`: debe ser `True`.


**Base proporcionada.** Datos del ejemplo; ejecuta esta celda antes de completar los huecos.

```python
book = [{'id': 1, 'size': 0.5}, {'id': 2, 'size': 0.2}]
```

```python
def cancel_order(book, order_id):
    result = []
    for order in book:
        if order['id'] != order_id:
            result.append(order)
    return result

def cancel_compact(book, order_id):
    return [order for order in book if order['id'] != order_id]

new_list = cancel_order(book, 1)
same_order = new_list[0] is book[1]
```

```python
print(cancel_order(book, 1))
print(cancel_compact(book, 1))
print(new_list, same_order)
print(book)
```

**Resultado:** Ambas funciones y `new_list`: `[{'id': 2, 'size': 0.2}]`; `same_order`: `True`. El libro original conserva las dos órdenes.

### Ampliación · Volumen de cada lado

Completa `imbalance(book)` sumando tamaños de compras y ventas con
generadores dentro de `sum` y devolviendo `(compras - ventas) / (compras + ventas)`.
Los tamaños son no negativos; si el volumen total es cero, devuelve `None`.
Para el libro dado debe resultar `0.5`.


**Base proporcionada.** Datos del ejemplo; ejecuta esta celda antes de completar los huecos.

```python
book = [{'side': 'buy', 'size': 2},
    {'side': 'sell', 'size': 1},
    {'side': 'buy', 'size': 1}]
```

```python
def imbalance(book):
    buys = sum(order['size'] for order in book if order['side'] == 'buy')
    sells = sum(order['size'] for order in book if order['side'] == 'sell')
    total = buys + sells
    if total == 0:
        return None
    return (buys - sells) / total
```

```python
print(imbalance(book))
print(imbalance([]))
print(imbalance([{'side': 'buy', 'size': 4}]))
print(imbalance([{'side': 'sell', 'size': 4}]))
```

**Resultado:** `0.5`, `None`, `1.0`, `-1.0`.

### Ampliación · Dos formas de dar un criterio

Define `get_price(order)` y pásala como `key` a `sorted` para guardar las
órdenes de menor a mayor precio en `ordered`.
Repite con una lambda en `ordered_lambda`: ambas listas deben tener ids
`['B', 'A', 'C']` y `orders` debe conservar su orden original.


**Base proporcionada.** Datos del ejemplo; ejecuta esta celda antes de completar los huecos.

```python
orders = [{'id': 'A', 'price': 105},
      {'id': 'B', 'price': 100},
      {'id': 'C', 'price': 110}]
```

```python
def get_price(order):
    return order['price']

ordered = sorted(orders, key=get_price)
ordered_lambda = sorted(orders, key=lambda order: order['price'])
```

```python
print([order['id'] for order in ordered])
print([order['id'] for order in ordered_lambda])
print([order['id'] for order in orders])
```

**Resultado:** Las dos listas nuevas: `['B', 'A', 'C']`; la original: `['A', 'B', 'C']`.

donde un punto básico es `1/10000`.

```python
def fee(notional, bps=1):
    return notional * bps / 10000
```

```python
print(fee(100000))
print(fee(100000, 5))
print(fee(20000, bps=0))
```
