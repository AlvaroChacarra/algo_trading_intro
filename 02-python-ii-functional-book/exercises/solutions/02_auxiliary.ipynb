### Filtra sin cambiar la lista

Implementa `cancel_order(book, order_id)` con `for` y `append`, y
`cancel_compact` con una comprensión: ambas devuelven una lista nueva sin ese id,
conservando el orden y los diccionarios originales.
Guarda en `new_list` la cancelación del id `1` y en `same_order` si su único
elemento es el mismo objeto que `book[1]`: debe ser `True`.


```python
book = [{'id': 1, 'size': 0.5}, {'id': 2, 'size': 0.2}]
def cancel_order(book, order_id):
    result = []
    for order in book:
        if order['id'] != order_id:
            result.append(order)
    return result

def cancel_compact(book, order_id):
    return [order for order in book if order['id'] != order_id]

new_list = cancel_order(book, 1)
same_order = new_list[0] is book[1]
```

```python
print(cancel_order(book, 1))
print(cancel_compact(book, 1))
print(new_list, same_order)
print(book)
```

**Resultado esperado:** Ambas funciones y `new_list`: `[{'id': 2, 'size': 0.2}]`; `same_order`: `True`. El libro original conserva las dos órdenes.

### Volumen de cada lado

Completa `imbalance(book)` sumando tamaños de compras y ventas con
generadores dentro de `sum` y devolviendo `(compras - ventas) / (compras + ventas)`.
Los tamaños son no negativos; si el volumen total es cero, devuelve `None`.
Para el libro dado debe resultar `0.5`.


```python
book = [{'side': 'buy', 'size': 2},
    {'side': 'sell', 'size': 1},
    {'side': 'buy', 'size': 1}]
def imbalance(book):
    buys = sum(order['size'] for order in book if order['side'] == 'buy')
    sells = sum(order['size'] for order in book if order['side'] == 'sell')
    total = buys + sells
    if total == 0:
        return None
    return (buys - sells) / total
```

```python
print(imbalance(book))
print(imbalance([]))
print(imbalance([{'side': 'buy', 'size': 4}]))
print(imbalance([{'side': 'sell', 'size': 4}]))
```

**Resultado esperado:** `0.5`, `None`, `1.0`, `-1.0`.

### Dos formas de dar un criterio

Define `get_price(order)` y pásala como `key` a `sorted` para guardar las
órdenes de menor a mayor precio en `ordered`.
Repite con una lambda en `ordered_lambda`: ambas listas deben tener ids
`['B', 'A', 'C']` y `orders` debe conservar su orden original.


```python
orders = [{'id': 'A', 'price': 105},
      {'id': 'B', 'price': 100},
      {'id': 'C', 'price': 110}]
def get_price(order):
    return order['price']

ordered = sorted(orders, key=get_price)
ordered_lambda = sorted(orders, key=lambda order: order['price'])
```

```python
print([order['id'] for order in ordered])
print([order['id'] for order in ordered_lambda])
print([order['id'] for order in orders])
```

**Resultado esperado:** Las dos listas nuevas: `['B', 'A', 'C']`; la original: `['A', 'B', 'C']`.

### Una comisión por defecto

donde un punto básico es `1/10000`.
`fee(100000)` debe devolver `10` y `fee(100000, 5)` debe devolver `50`.
