## A1. Spread y mid

**Practicas:** componer funciones.

Usando best_bid/best_ask, escribe `spread(book)` y `mid(book)`.

````python
def best_bid(book):
    return max(o['price'] for o in book if o['side']=='buy')
def best_ask(book):
    return min(o['price'] for o in book if o['side']=='sell')
book = [{'side':'buy','price':100},{'side':'sell','price':102}]
def spread(book):
    pass
def mid(book):
    pass
````

````python
assert spread(book) == 2
assert mid(book) == 101
print('ok')
````

### Solución guiada

```python
def spread(book):
    return best_ask(book) - best_bid(book)
def mid(book):
    return (best_bid(book) + best_ask(book)) / 2
```

## A2. Nocional total

**Practicas:** acumular sobre el libro.

Escribe `total_notional(book)` = suma de price*size de todas las órdenes.

````python
book = [{'price':100,'size':0.5},{'price':200,'size':0.25}]
def total_notional(book):
    pass
````

````python
assert abs(total_notional(book) - 100) < 1e-9
print('ok')
````

### Solución guiada

```python
def total_notional(book):
    return sum(o['price'] * o['size'] for o in book)
```
