### A1. Lados con seguridad

Haz que `Order.__init__` lance `ValueError` si `side` no es 'buy' ni 'sell'. Un objeto puede proteger sus propios datos.

<sub>practicas: validar en __init__</sub>

````python
class Order:
    def __init__(self, symbol, side, price, size):
        pass
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
try:
    Order('X','byu',1,1); raise SystemExit('deberia fallar')
except ValueError:
    pass
assert Order('X','buy',1,1).side == 'buy'
print('ok')
````

<details>
<summary>💡 Ver solución</summary>

```python
class Order:
    def __init__(self, symbol, side, price, size):
        if side not in ('buy','sell'):
            raise ValueError('side debe ser buy o sell')
        self.symbol=symbol; self.side=side; self.price=price; self.size=size
```

</details>
