### A7. La caja tras la sesión

Con la lista `fills`, suma todos los `cash_flow()` en `total_cash`.

<sub>practicas: objetos en una lista</sub>

````python
class Fill:
    def __init__(self, side, price, size):
        self.side = side; self.price = price; self.size = size
    def cash_flow(self):
        return -self.price * self.size if self.side == 'buy' else self.price * self.size

fills = [Fill('buy', 99950, 0.5), Fill('sell', 100050, 0.5)]
total_cash = None
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert total_cash is not None, '⏸ total_cash sigue en None: completa el ejercicio antes de validar'
assert abs(total_cash - 50.0) < 1e-9, 'compraste a 99950 y vendiste a 100050'
print('ok ->', total_cash)
````

<details>
<summary>💡 Ver solución</summary>

```python

for f in fills:
    total_cash = total_cash + f.cash_flow()
```

### A11. Lados con seguridad

Haz que `Order.__init__` lance `ValueError` si `side` no es 'buy' ni 'sell'. Un objeto puede proteger sus propios datos.

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
try:
    Order('X','byu',1,1); raise SystemExit('deberia fallar')
except ValueError:
    pass
assert Order('X','buy',1,1).side == 'buy'
print('ok')
````
