### Tu Fill · una sola celda editable

Completa los huecos 1 y 3 siguiendo los pasos de debajo. Al editar la clase, vuelve a ejecutarla y crea instancias nuevas: las anteriores conservan su definición.

**Base proporcionada.** Registro con side, price y size, como en la presentación. isfinite mantiene la validación proporcionada; no es el objetivo de la práctica.

```python
from math import isfinite

fill_record = {'side': 'buy', 'price': 100, 'size': 2}
```

```python
class Fill:
    def __init__(self, side, price, size):
        valid_numbers = all(isfinite(x) and x > 0 for x in (price, size))
        if side not in ('buy', 'sell') or not valid_numbers:
            raise ValueError('fill requiere lado válido y valores positivos finitos')
        self.side = side
        self.price = price
        self.size = size

    def cash_flow(self):
        sign = -1 if self.side == 'buy' else 1
        return sign * self.price * self.size
```

#### 1 · Del registro al objeto

Completa las tres asignaciones del constructor. Ejecuta la clase y crea la compra desde el registro conocido; observa sus atributos.

```python
f = Fill(fill_record['side'], fill_record['price'], fill_record['size'])
print('compra:', f.side, f.price, f.size)
```

**Resultado:** `buy 100 2`. Una orden era intención; este registro y este objeto describen una ejecución confirmada.

#### 2 · Dos instancias independientes

Antes de crear la venta, predice si cambiarán los datos de f. Ejecuta y compara ambos objetos.

```python
g = Fill('sell', 110, 2)
print('venta:', g.side, g.price, g.size)
print('compra intacta:', f.side, f.price, f.size)
print('objetos distintos:', f is not g)
```

**Resultado:** `sell 110 2`, `buy 100 2` y `True`.

#### 3 · El objeto calcula su flujo

Completa cash_flow. Reejecuta la clase y las dos observaciones anteriores para crear instancias nuevas; luego predice el signo de cada llamada.

```python
print('flujo compra:', f.cash_flow())
print('flujo venta:', g.cash_flow())
print('otra consulta:', f.cash_flow())
```

**Resultado:** `-200`, `220`, `-200`. Volver a consultar no registra otro fill.

### 4 · Reúne compra y venta

Guarda en total_cash la suma de los flujos de tus dos objetos. Predice la caja neta antes de ejecutar; no cambies precio ni tamaño para introducir el signo.

```python
total_cash = f.cash_flow() + g.cash_flow()
```

```python
print('caja de la vuelta completa:', total_cash)
print('cantidad neta:', f.size - g.size)
```

**Resultado:** `20` y `0`: dos ejecuciones sin comisiones cierran la posición.

### 5 · Vende solo una unidad

Crea partial_sell a 110 por una unidad. Calcula partial_cash sumando su flujo a la compra inicial y remaining_units restando tamaños; predice si la posición queda cerrada.

```python
partial_sell = Fill('sell', 110, 1)
partial_cash = f.cash_flow() + partial_sell.cash_flow()
remaining_units = f.size - partial_sell.size
```

```python
print('venta parcial: caja / unidades:', partial_cash, remaining_units)
```

**Resultado:** `-90 1`: queda inventario; esa caja por sí sola no mide el resultado total.
