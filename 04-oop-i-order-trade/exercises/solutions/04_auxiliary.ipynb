### Variante · Dos fills independientes

Crea f y g como compras de 2 unidades a 100. Cambia solo f.price a 110; guarda sus flujos en flows y predice si ambos cambian.

**Base proporcionada.** Fill resuelto del principal, generado desde su misma respuesta.

```python
from math import isfinite

fill_record = {'side': 'buy', 'price': 100, 'size': 2}
class Fill:
    def __init__(self, side, price, size):
        valid_numbers = all(isfinite(x) and x > 0 for x in (price, size))
        if side not in ('buy', 'sell') or not valid_numbers:
            raise ValueError('fill requiere lado válido y valores positivos finitos')
        self.side = side
        self.price = price
        self.size = size

    def cash_flow(self):
        sign = -1 if self.side == 'buy' else 1
        return sign * self.price * self.size
```

```python
f = Fill('buy', 100, 2)
g = Fill('buy', 100, 2)
f.price = 110
flows = [f.cash_flow(), g.cash_flow()]
```

```python
print('flujos independientes:', flows)
print('precios:', f.price, g.price)
```

**Resultado:** `[-220, -200]` y `110 100`. Esta mutación es un experimento sobre objetos; no reescribe una ejecución real.

### Variante · Un valor que queda antiguo

Completa el constructor guardando `side`, `price`, `size` y el producto
en `self.notional` como atributo.
Para `o`, guarda `stored_before`, cambia su precio a `110` y guarda
`stored_after`, `recomputed` y `stale` comparando el atributo con el producto actual.
Predice los cuatro resultados antes de ejecutar.


```python
class OrderDrill:
    def __init__(self, side, price, size):
        self.side = side
        self.price = price
        self.size = size
        self.notional = price * size

o = OrderDrill('buy', 100, 2)
stored_before = o.notional
o.price = 110
stored_after = o.notional
recomputed = o.price * o.size
stale = stored_after != recomputed
```

```python
print('guardado antes:', stored_before)
print('guardado después:', stored_after)
print('producto actual:', recomputed)
print('se ha quedado antiguo:', stale)
```

**Resultado:** `200`, `200`, `220`, `True`: cambiar price no recalcula el atributo notional.

### Ampliación · Una clase, dos quotes

Completa `Quote`: guarda `bid` y `ask`, calcula su promedio en `mid()` y
devuelve `Quote(bid/ask)` en `__repr__` usando los valores actuales.
Crea `q1=Quote(99, 101)` y `q2=Quote(200, 204)`, y cambia `q1.bid` a `100`.
Deben quedar mids `100.5` y `202`, y `repr(q1)` debe ser `'Quote(100/101)'`.


```python
class Quote:
    def __init__(self, bid, ask):
        self.bid = bid
        self.ask = ask

    def mid(self):
        return (self.bid + self.ask) / 2

    def __repr__(self):
        return f'Quote({self.bid}/{self.ask})'

q1 = Quote(99, 101)
q2 = Quote(200, 204)
q1.bid = 100
```

```python
print('q1:', repr(q1), 'mid:', q1.mid())
print('q2:', repr(q2), 'mid:', q2.mid())
```

**Resultado:** `Quote(100/101)` y mid `100.5`; `Quote(200/204)` y mid `202.0`. Cambiar q1 no cambia q2.

### Ampliación · Rechaza un precio inválido

Completa `OrderDrill.__init__` para guardar `side`, `price` y `size`
cuando `price > 0`.
Si el precio es cero o negativo, debe lanzar `ValueError` antes de guardar
los atributos; `OrderDrill('buy', 100, 2)` debe conservar esos tres datos.


```python
class OrderDrill:
    def __init__(self, side, price, size):
        if price <= 0:
            raise ValueError(f'price inválido: {price}')
        self.side = side
        self.price = price
        self.size = size
```

```python
valid = OrderDrill('buy', 100, 2)
print('orden válida:', valid.side, valid.price, valid.size)
for price in (0, -5):
    try:
        OrderDrill('buy', price, 1)
        print('precio aceptado:', price)
    except ValueError:
        print('precio rechazado:', price)
```

**Resultado:** `buy 100 2`; después, `precio rechazado: 0` y `precio rechazado: -5`.
