### 6. De la orden al dinero

<sub>🔵 si vamos bien · ~6 min</sub>

Modela una vuelta completa: compras 0.5 @ 100 y luego vendes 0.5 @ 110. Crea los dos `Fill` y guarda `total_cash` = suma de sus `cash_flow()`.

> 💡 Compra resta (-50), venta suma (+55) -> total +5.

<sub>practicas: juntar Order, Fill y cash_flow</sub>

````python
class Fill:
    def __init__(self, symbol, side, price, size):
        self.symbol=symbol; self.side=side; self.price=price; self.size=size
    def cash_flow(self):
        return (-1 if self.side=='buy' else 1)*self.price*self.size
buy = None
sell = None
total_cash = None
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert buy is not None, '⏸ buy sigue en None: completa el ejercicio antes de validar'
assert sell is not None, '⏸ sell sigue en None: completa el ejercicio antes de validar'
assert total_cash is not None, '⏸ total_cash sigue en None: completa el ejercicio antes de validar'
assert abs(total_cash - 5.0) < 1e-9, 'compra -50, venta +55 -> +5'
print('ok  total_cash=%.1f' % total_cash)
````

<summary>💭 Pista (antes de mirar la solución)</summary>
