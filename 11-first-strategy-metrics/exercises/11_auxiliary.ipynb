### A1. Curva de equity

Corre la estrategia y guarda `curve = result.equity_curve`. Verifica que tiene 500 puntos. (En clase la dibujarías con matplotlib.)

<sub>practicas: serie temporal de PnL</sub>

````python
from exchange import Strategy, NewOrder, Order, Side, OrderType, Market, Backtest
class S(Strategy):
    def on_book_update(self, book):
        return [NewOrder(Order('BTCUSDT', Side.BUY, 0.01, order_type=OrderType.MARKET))]
curve = None
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert len(curve) == 500
print('ok  puntos=%d' % len(curve))
````
