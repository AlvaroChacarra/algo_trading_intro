## 4. La clase Fill

**Practicas:** segunda clase + método.

Define `Fill(symbol, side, price, size)` con método `cash_flow()` (buy negativo, sell positivo).

````python
class Fill:
    def __init__(self, symbol, side, price, size):
        pass
    def cash_flow(self):
        pass
````

````python
assert abs(Fill('X','buy',100,0.5).cash_flow() + 50) < 1e-9
assert abs(Fill('X','sell',100,0.5).cash_flow() - 50) < 1e-9
print('ok')
````

### Solución guiada

```python
class Fill:
    def __init__(self, symbol, side, price, size):
        self.symbol=symbol; self.side=side; self.price=price; self.size=size
    def cash_flow(self):
        sign = -1 if self.side=='buy' else 1
        return sign * self.price * self.size
```
