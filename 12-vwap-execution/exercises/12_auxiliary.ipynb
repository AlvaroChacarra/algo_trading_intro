<details>
<summary>💡 Ver solución</summary>

```python
def slope(xs, ys):
    n = len(xs); mx = sum(xs)/n; my = sum(ys)/n
    cov = sum((x-mx)*(y-my) for x,y in zip(xs,ys))
    var = sum((x-mx)**2 for x in xs)
    return cov / var
```

## 🏋️ Transferencia — la misma idea, otro dominio

El VWAP no inventa nada: reparte un total en el tiempo siguiendo un perfil de actividad. La misma cuenta sirve para una panadería que hornea según la demanda por hora.

### T1. La panadería que sigue la demanda (VWAP sin mercado)

<sub>🟣 bonus · ~5 min</sub>

Una panadería quiere hornear **100 barras** repartidas en una jornada de 8 horas, proporcionalmente a la demanda esperada `demanda` (clientes por hora). Calcula `plan[i]` = barras a hornear en la hora i: normaliza `demanda` para que sume 1 y multiplica por 100. Es exactamente el perfil VWAP.

<sub>practicas: perfil normalizado fuera del trading</sub>

````python
demanda = [5, 10, 20, 25, 20, 10, 6, 4]  # clientes por hora
plan = None
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert plan is not None, '⏸ plan sigue en None: completa el ejercicio antes de validar'
assert abs(sum(plan) - 100) < 1e-9, 'el plan reparte las 100 barras enteras'
assert plan.index(max(plan)) == demanda.index(max(demanda)), 'horneas más donde hay más demanda'
print('ok — el VWAP era repartir un total según un perfil; el pan no lo sabe, pero lo hace igual')
````
