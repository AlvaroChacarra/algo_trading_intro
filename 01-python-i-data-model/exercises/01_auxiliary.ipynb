### A2. Mejor bid y mejor ask

De `book` (lista de órdenes), saca `best_bid` (precio de compra más alto) y `best_ask` (precio de venta más bajo).

> 💡 Filtra por `side` dentro de un `max`/`min` con generador.

<sub>practicas: max / min con filtro</sub>

````python
book = [
    {'side':'buy','price':99980,'size':0.1}, {'side':'buy','price':99990,'size':0.2},
    {'side':'sell','price':100010,'size':0.15}, {'side':'sell','price':100005,'size':0.1},
]
best_bid = None
best_ask = None
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert best_bid == 99990, 'best_bid debe ser 99990'
assert best_ask == 100005, 'best_ask debe ser 100005'
print('ok')
````

<details>
<summary>💡 Ver solución</summary>

```python
best_bid = max(o['price'] for o in book if o['side']=='buy')
best_ask = min(o['price'] for o in book if o['side']=='sell')
```

</details>

### A5. Ver el bytecode con dis

Python compila tu texto a bytecode. Compila `'mid = (bid + ask) / 2'` y captura su desensamblado en `bytecode` (string). Comprueba que contiene instrucciones de la VM.

> 💡 `dis.dis(compile(src, '<x>', 'exec'), file=buf)` escribe el bytecode en un buffer.

<sub>practicas: compilar texto a bytecode</sub>

````python
import dis, io
src = 'mid = (bid + ask) / 2'
bytecode = None
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert isinstance(bytecode, str) and 'LOAD_NAME' in bytecode, 'debe contener instrucciones como LOAD_NAME'
print('ok — esto es lo que ejecuta la máquina virtual de Python')
````

<details>
<summary>💡 Ver solución</summary>

```python

dis.dis(compile(src, '<x>', 'exec'), file=buf)
bytecode = buf.getvalue()
print(bytecode)
```
