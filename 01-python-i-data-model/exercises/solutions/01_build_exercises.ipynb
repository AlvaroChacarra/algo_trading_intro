### Una foto del mercado

1. Completa `spread` y `mid` a partir de `bid` y `ask`. Predice qué cambia si ask pasa a 103; prueba y vuelve a los datos iniciales.

2. El bucle ya recorre dos diccionarios `quote`. Completa los cálculos y las etiquetas: spread ≤ 2 es `"estrecho"`; si no, `"amplio"`. `append` añade un resumen a `results` en cada vuelta.

No necesitas definir funciones. Compara los dos resúmenes y ejecuta todas las celdas desde un kernel nuevo.


```python
bid = 99
ask = 101
quotes = [
    {"bid": 99, "ask": 101},
    {"bid": 99, "ask": 103},
]
spread = ask - bid
mid = (bid + ask) / 2

results = []
for quote in quotes:
    quote_spread = quote["ask"] - quote["bid"]
    quote_mid = (quote["bid"] + quote["ask"]) / 2
    if quote_spread <= 2:
        state = "estrecho"
    else:
        state = "amplio"
    results.append({"spread": quote_spread, "mid": quote_mid, "state": state})
```

```python
print('spread:', spread)
print('mid:', mid)
print('resúmenes:', results)
```

**Resultado esperado:** `99/101` → spread `2`, mid `100`. Resúmenes de las dos quotes: `2, 100, estrecho` y `4, 101, amplio`. Al cambiar un dato, ejecuta desde el inicio.
