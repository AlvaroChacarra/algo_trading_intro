### Una lista que crece

Añade `102` al final de `mids` y guarda `first_mid`, `last_mid` y `n_mids`.
Después guarda `first2` y `last3` con los dos primeros y los tres últimos precios:
deben ser `[99, 100]` y `[98, 101, 102]`.


```python
mids = [99, 100, 98, 101]
mids.append(102)
first_mid = mids[0]
last_mid = mids[-1]
n_mids = len(mids)
first2 = mids[:2]
last3 = mids[-3:]
```

```python
print(mids)
print(first_mid, last_mid, n_mids)
print(first2, last3)
```

**Resultado esperado:** `[99, 100, 98, 101, 102]`; `99 102 5`; `[99, 100] [98, 101, 102]`.

### Suma y media

Recorre `mids` con un `for` para calcular `total` y después `average`.
Comprueba la suma con `total_sum = sum(mids)` y guarda también `lowest` y `highest`:
la suma debe ser `404`, la media `101`, el mínimo `99` y el máximo `104`.


```python
mids = [99, 101, 100, 104]
total = 0
for mid in mids:
    total = total + mid
average = total / len(mids)
total_sum = sum(mids)
lowest = min(mids)
highest = max(mids)
```

```python
print(total, total_sum, average)
print(lowest, highest)
```

**Resultado esperado:** `404 404 101.0`; `99 104`.

### Consulta y actualiza posiciones

Antes de modificar `positions`, guarda `btc_pos` y `eth_pos` usando `get`
con default `0`, y `had_eth` comprobando si existe la clave `'ETHUSDT'`.
Después suma `0.2` a BTC y añade `1.0` ETH: los valores iniciales deben ser
`0.5`, `0` y `False`, y las posiciones finales `0.7` y `1.0`.


```python
positions = {'BTCUSDT': 0.5}
btc_pos = positions.get('BTCUSDT', 0)
eth_pos = positions.get('ETHUSDT', 0)
had_eth = 'ETHUSDT' in positions
positions['BTCUSDT'] = btc_pos + 0.2
positions['ETHUSDT'] = eth_pos + 1.0
```

```python
print(btc_pos, eth_pos, had_eth)
print(positions)
```

**Resultado esperado:** `0.5 0 False`; después `{'BTCUSDT': 0.7, 'ETHUSDT': 1.0}`.

### Clasifica cuatro spreads

Recorre `spreads` y añade a `market_states` una clasificación por valor:
`'tight'` hasta `20`, `'normal'` por encima de `20` y hasta `60`, o `'wide'` si supera `60`.
Debes obtener `['tight', 'normal', 'normal', 'wide']` usando `if`, `elif` y `else`.


```python
spreads = [20, 21, 60, 61]
market_states = []
for spread in spreads:
    if spread <= 20:
        market_state = 'tight'
    elif spread <= 60:
        market_state = 'normal'
    else:
        market_state = 'wide'
    market_states.append(market_state)
```

```python
print(market_states)
```

**Resultado esperado:** `['tight', 'normal', 'normal', 'wide']`.
