### 5. Mide la presión

Recorre `book` y suma `buy_volume` y `sell_volume` por lado.

<sub>practicas: for e if</sub>

````python
book = [
    {'side': 'buy',  'price': 99980, 'size': 0.10},
    {'side': 'sell', 'price': 100010, 'size': 0.15},
    {'side': 'buy',  'price': 99970, 'size': 0.20},
]
buy_volume = 0
sell_volume = 0
# recorre book
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert abs(buy_volume - 0.30) < 1e-9, 'buy_volume debe ser 0.30'
assert abs(sell_volume - 0.15) < 1e-9, 'sell_volume debe ser 0.15'
print('ok')
````

<details>
<summary>💡 Ver solución</summary>

```python

sell_volume = 0
for o in book:
    if o['side'] == 'buy':
        buy_volume += o['size']
    else:
        sell_volume += o['size']
```
