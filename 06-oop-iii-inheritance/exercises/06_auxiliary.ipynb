### C2. mid() exprés

Completa `mid()` apoyándote en los otros dos métodos.

<sub>practicas: repaso: métodos compuestos</sub>

````python
class OrderBook:
    def __init__(self, bids, asks):
        self.bids = bids; self.asks = asks
    def best_bid(self):
        return max(p for p, s in self.bids)
    def best_ask(self):
        return min(p for p, s in self.asks)
class OrderBook(OrderBook):
    def mid(self):
        pass

ob = OrderBook([(99950, 0.5)], [(100000, 0.3)])
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert ob.mid() == 99975.0
print('ok')
````

<details>
<summary>💡 Ver solución</summary>

```python
class OrderBook(OrderBook):
    def mid(self):
        return (self.best_bid() + self.best_ask()) / 2

ob = OrderBook([(99950, 0.5)], [(100000, 0.3)])
```

</details>

Una llamada, muchas respuestas — y la letra pequeña honesta.

### A5. El bucle polimórfico

Monta `familia = [Strategy('b'), Momentum('m'), Contraria('c')]` y construye `decisions` llamando `s.decide(0.5)` a cada una en un bucle.

<sub>practicas: misma llamada, respuestas distintas</sub>

````python
class Strategy:
    def __init__(self, name):
        self.name = name
    def decide(self, imbalance):
        return 'hold'

class Momentum(Strategy):
    def decide(self, imbalance):
        if imbalance > 0.3:
            return 'buy'
        if imbalance < -0.3:
            return 'sell'
        return 'hold'

class Contraria(Strategy):
    def decide(self, imbalance):
        if imbalance > 0.3:
            return 'sell'
        if imbalance < -0.3:
            return 'buy'
        return 'hold'
familia = None
decisions = None
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert decisions == ['hold', 'buy', 'sell'], 'tres clases, tres respuestas a la MISMA llamada'
print('ok ->', decisions)
````

<summary>💡 Ver solución</summary>

```python

    decisions.append(s.decide(0.5))
```

### A7. La impostora (duck typing)

`Impostora` NO hereda de `Strategy` pero tiene `decide`. Guarda `llama = Impostora().decide(0.9)` y `es_strategy = isinstance(Impostora(), Strategy)`. Python permite lo primero aunque lo segundo sea False — el bucle funciona por el **método**, no por la familia.

<sub>practicas: la letra pequeña honesta</sub>

### A8. Firma el contrato

Con la ABC dada, escribe `Completa(Strategy)` que implemente `decide` devolviendo `'hold'`. Instánciala — ahora sí se puede.

<sub>practicas: implementar el abstracto</sub>

````python
from abc import ABC, abstractmethod

class Strategy(ABC):
    @abstractmethod
    def decide(self, imbalance): ...
c = None
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert c.decide(0.5) == 'hold'
print('ok — con decide implementado, la ABC te deja pasar')
````

### A9. Contrato + super

La ABC dada guarda `name` en su `__init__`. Escribe `Mini(Strategy)` que use `super().__init__(name)` y decida `'buy'` si `imbalance > 0`, si no `'sell'`.

<sub>practicas: ABC con __init__ común</sub>

````python
from abc import ABC, abstractmethod

class Strategy(ABC):
    def __init__(self, name):
        self.name = name
    @abstractmethod
    def decide(self, imbalance): ...
class Mini(Strategy):
    pass

m = None
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert m.name == 'm1'
assert m.decide(0.2) == 'buy' and m.decide(-0.2) == 'sell'
print('ok')
````
