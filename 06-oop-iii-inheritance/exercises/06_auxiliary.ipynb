### A1. isinstance y la familia

Con `Momentum(Strategy)`, comprueba que una instancia es a la vez `Momentum` y `Strategy`. Guarda `es_momentum` y `es_strategy`.

<sub>practicas: comprobar el tipo</sub>

````python
class Strategy:
    pass
class Momentum(Strategy):
    pass
m = Momentum()
es_momentum = None
es_strategy = None
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert es_momentum is True and es_strategy is True
print('ok')
````

### A2. La ABC obliga a implementar

Una subclase que NO implementa el método abstracto tampoco se puede instanciar. Comprueba que instanciar `Incompleta()` lanza `TypeError`.

<sub>practicas: por qué sirve @abstractmethod</sub>

````python
from abc import ABC, abstractmethod
class Strategy(ABC):
    @abstractmethod
    def decide(self, imbalance): ...
class Incompleta(Strategy):
    pass
fallo = None  # ponlo a True si Incompleta() lanza TypeError
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert fallo is True
print('ok — la ABC te protege de subclases a medio hacer')
````

<details>
<summary>💡 Ver solución</summary>

```python
try:
    Incompleta()
    fallo = False
except TypeError:
    fallo = True
```
