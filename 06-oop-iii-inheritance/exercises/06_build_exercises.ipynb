### 2. Hereda y sobrescribe

Define `AlwaysBuy(Strategy)` que sobrescriba `decide` para devolver siempre `'buy'`. Debe seguir siendo una `Strategy`.

<sub>practicas: herencia + override</sub>

````python
class Strategy:
    def decide(self, imbalance):
        return 'hold'
class AlwaysBuy(Strategy):
    pass
````

### 3. El contrato: clase abstracta

Haz `Strategy` **abstracta** con `@abstractmethod` en `decide`. Ya no se podrá instanciar `Strategy()` directamente; solo subclases que implementen `decide`.

> 💡 `from abc import ABC, abstractmethod` y `class Strategy(ABC)`.

<sub>practicas: ABC + @abstractmethod</sub>

````python
from abc import ABC, abstractmethod

class Strategy(ABC):
    pass

class Buyer(Strategy):
    def decide(self, imbalance):
        return 'buy'
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
try:
    Strategy(); raise SystemExit('no deberia poder instanciarse')
except TypeError:
    pass
assert Buyer().decide(0) == 'buy'
print('ok')
````

<details>
<summary>💡 Ver solución</summary>

```python
from abc import ABC, abstractmethod

class Strategy(ABC):
    @abstractmethod
    def decide(self, imbalance):
        ...

class Buyer(Strategy):
    def decide(self, imbalance):
        return 'buy'
```
