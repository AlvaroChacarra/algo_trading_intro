### A5. El barrido de γ, tuyo

<sub>practicas: la frontera riesgo/retorno</sub>

````python
from exchange.simulation import MMSimulation
from exchange.strategies.market_maker import AvellanedaStoikov

def corre(gamma):
    mm = AvellanedaStoikov('SIM', quote_size=0.1, gamma=gamma, sigma=0.5, horizon=50)
    return MMSimulation(mm, steps=50, seed=3).run()
inv_low = None
inv_high = None
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert inv_low >= inv_high, 'mas gamma = menos inventario maximo (o igual)'
print(f'ok  g=0.1 -> {inv_low:.2f} | g=1.0 -> {inv_high:.2f}')
````
