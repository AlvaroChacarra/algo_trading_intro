## A2. PositionTracker del paquete

**Practicas:** usar exchange.

Usa `PositionTracker` y `Fill` de `exchange`: aplica un fill de compra y mira equity.

````python
from exchange import PositionTracker
from exchange.trades import Fill
tracker = PositionTracker()
# aplica un Fill de compra de 0.5 @ 100
````

````python
assert abs(tracker.equity(100) - 0) < 1e-6, 'comprar a mid no cambia equity'
assert tracker.position > 0
print('ok')
````

### Solución guiada

```python
from exchange import PositionTracker
from exchange.trades import Fill
tracker = PositionTracker()
tracker.apply_fill(Fill(1, 'BTCUSDT', 'buy', 100, 0.5))
```
