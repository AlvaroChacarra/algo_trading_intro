### A7. ¿Predice el imbalance?

<sub>🟣 OPTIONAL · profundización no evaluable · ~5 min</sub>

Cuenta cuántas veces, cuando `imbalance(1) > 0`, el mid del siguiente snapshot sube. Guarda el conteo en `up_after_positive`.

<sub>practicas: señal vs siguiente mid</sub>

````python
import csv
import os
import exchange

_data_path = os.path.join(os.path.dirname(exchange.__file__), '_data', 'btc_lob_snapshots.csv')
with open(_data_path, newline='') as _f:
    row = {k: float(v) for k, v in next(csv.DictReader(_f)).items()}
from exchange.book import OrderBook
with open(_data_path, newline='') as _f:
    rows = [{k: float(v) for k, v in r.items()} for r in csv.DictReader(_f)]
books = [OrderBook.from_snapshot('BTCUSDT', r) for r in rows]
up_after_positive = 0
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert up_after_positive != 0, 'sigue en 0: recorre el dia comparando cada mid con el anterior'
assert up_after_positive == 117, f'obtuviste {up_after_positive}; cuenta solo cuando imbalance(1) > 0 Y el mid del snapshot siguiente sube'
print('ok  veces que sube tras imbalance+:', up_after_positive)
````

<details>
<summary>💡 Ver solución</summary>

```python

for prev, cur in zip(books, books[1:]):
    if prev.imbalance(1) is not None and prev.imbalance(1) > 0 and cur.mid > prev.mid:
        up_after_positive += 1
```
