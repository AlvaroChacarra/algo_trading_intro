### 2. El skew baja las cotizaciones si estás largo

<sub>🟢 LIVE · núcleo presencial · ~5 min</sub>

Con el mismo MM, fija `mm.inventory = 2` y vuelve a cotizar. El nuevo `bid2` debe ser menor que el bid de inventario 0.

<sub>practicas: inventario -> reservation price</sub>

````python
from exchange import OrderBook, Level
from exchange.strategies import MarketMaker
book = OrderBook('BTC', [Level(99.5,1)], [Level(100.5,1)])
mm = MarketMaker('BTC', half_spread=0.5, inventory_skew=1.0)
bid0, _ = mm.quotes(book)
bid2 = None
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert bid2 is not None, '⏸ bid2 sigue en None: completa el ejercicio antes de validar'
assert bid2 < bid0, 'largo de inventario -> cotizas más bajo para soltar'
print('ok  bid0=%.2f bid2=%.2f' % (bid0, bid2))
````
