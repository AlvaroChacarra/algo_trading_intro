````python
fills = [(99.4, 0.1), (100.6, 0.1)]
avg = None
````

````python
# ✅ Comprobación — ejecútala (Shift+Enter). Está plegada a propósito.
assert avg is not None, '⏸ avg sigue en None: completa el ejercicio antes de validar'
assert abs(avg - 100.0) < 1e-9
print('ok')
````
